```python
X_train=df[["Pclass","Age","SibSp","Parch","Fare","Sex_male","Embarked_C","Embarked_Q","Embarked_S"]]
y_train=df["Survived"]

X_test=df_test[["Pclass","Age","SibSp","Parch","Fare","Sex_male","Embarked_C","Embarked_Q","Embarked_S"]]

sonmodel = RandomForestClassifier(max_depth=5, random_state=42,class_weight='balanced')
sonmodel.fit(X_train, y_train)

```

```python
y_pred=sonmodel.predict(X_test)
```

```python
confusion_matrix(y_test, y_pred)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```
